```python
inventory={
    "apple":{"quantity":10,"price":30},
    "banana":{"quantity":15,"price":10},
    "milk":{"quantity":8,"price":120}
}
for item ,info in inventory.items():
    print(f"{ item.capitalize()}:Quantity={info['quantity']},Price={info['price']}Rs")
```

```python
#QUETSION NUMBER 3:
#Create a dictionary to store student grade and calculate the average grade.
```

```python
grades={
    "Ali":[85,75,90],
    "Sara":[92,88,95],
    "Bilal":[70,65,80]
}

#Calculating the average grade of eac student

for student ,marks in grades.items():
    avg=sum(marks)/len(marks)
    print(f"{student}'s average grade:{avg:.2f}")
```

```python
#QUESTION NUMBER 4:
#INVENTORY MANAGEMENT:
#Create a dictionary to store product information :name,quantity,price.
#Implementt functions to :
#Add new products......................Generate a report of low stock items
```

```python
inventory={
    "Laptop":{"quantity":5,"price":80000},
    "Mouse":{"quantity":15,"price":1000},
    "Keyboard":{"quantity":10,"price":2500}
}

#Add new product
inventory["Charger"] = {"quantity": 3, "price": 1500}
print("New product added:", "Charger")

#Update product quantity
inventory["Mouse"]["quantity"] = 8
print("Quantity of Mouse updated to:", inventory["Mouse"]["quantity"])

#  Calculate total inventory value 
total_value = 0
for item in inventory.values():
    total_value += item["quantity"] * item["price"]
print("Total Inventory Value = Rs", total_value)

# Generate report of low stock items (less than 5)
print("Low stock items:")
for name, item in inventory.items():
    if item["quantity"] < 5:
        print(f"- {name}: {item['quantity']} left")


```
